# ZlinkHost Joint Control Example

This notebook shows how to initialize `ZlinkHost`, read four joint angles, and move the four servos by joint angles in degrees.

- Install if needed: `pip install numpy pyserial`
- Joint range: `0..270` deg
- Servo pulse range: `500..2500`
- Default servo IDs: `0, 1, 2, 3`


```python
import numpy as np

from servo_host import ZlinkHost
```

```python
PORT = "COM3"

host = ZlinkHost(port=PORT)
print(f"Connected to {PORT}")
```

## Example 1: read_joints()

Read the current angles of servo IDs `0..3`.


```python
%timeit current_joints = host.read_joints()
```

```python
current_joints = host.read_joints()
print("Current joint angles (deg):", np.round(current_joints, 2))
```

## Example 2: move_to_joints()

Send a target angle array and a motion time in seconds.


```python
target_joints = np.array([190, 190, 190, 190], dtype=float)
move_time_s = 2.0

host.move_to_joints(target_joints, t=move_time_s)
print("Sent target joints (deg):", target_joints)
print("Move time (s):", move_time_s)
```

```python
updated_joints = host.read_joints()
print("Updated joint angles (deg):", np.round(updated_joints, 2))
```

## Optional: move back to a neutral pose


```python
neutral_joints = np.array([135.0, 135.0, 135.0, 135.0], dtype=float)
host.move_to_joints(neutral_joints, t=1.5)
print("Sent neutral joints (deg):", neutral_joints)
```

```python
host.close()
print("Serial port closed.")
```
